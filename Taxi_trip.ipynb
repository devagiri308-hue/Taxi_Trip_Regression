```python
# Mean
df['Trip_Distance_km'] = df['Trip_Distance_km'].fillna(df['Trip_Distance_km'].mean())

df['Passenger_Count'] = df['Passenger_Count'].fillna(df['Passenger_Count'].mean())

df['Base_Fare'] = df['Base_Fare'].fillna(df['Base_Fare'].mean())

df['Per_Km_Rate'] = df['Per_Km_Rate'].fillna(df['Per_Km_Rate'].mean())

df['Per_Minute_Rate'] = df['Per_Minute_Rate'].fillna(df['Per_Minute_Rate'].mean())

df['Trip_Duration_Minutes'] = df['Trip_Duration_Minutes'].fillna(df['Trip_Duration_Minutes'].mean())

df['Trip_Price'] = df['Trip_Price'].fillna(df['Trip_Price'].mean())

#Mode
df['Time_of_Day'] = df['Time_of_Day'].fillna(df['Time_of_Day'].mode()[0])

df['Day_of_Week'] = df['Day_of_Week'].fillna(df['Day_of_Week'].mode()[0])

df['Traffic_Conditions'] = df['Traffic_Conditions'].fillna(df['Traffic_Conditions'].mode()[0])

df['Weather'] = df['Weather'].fillna(df['Weather'].mode()[0])
```

```python
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm
```

```python
formula = 'Trip_Price ~ Trip_Distance_km + Passenger_Count + Base_Fare + Per_Km_Rate + Per_Minute_Rate + Trip_Duration_Minutes '
model = ols(formula, df).fit()
aov_table = anova_lm(model)
print(aov_table)
```

```python
formula = 'Trip_Price ~ C(Time_of_Day) + C(Day_of_Week) + C(Traffic_Conditions) + C(Weather)'
model = ols(formula, df).fit()
aov_table = anova_lm(model)
print(aov_table)
```

```python
df.drop(['Time_of_Day','Day_of_Week','Weather'],axis=1,inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder
LE = LabelEncoder()
df['Trip_Distance_km']=LE.fit_transform(df['Trip_Distance_km'])
df['Passenger_Count']=LE.fit_transform(df['Passenger_Count'])
df['Base_Fare']=LE.fit_transform(df['Base_Fare'])
df['Per_Km_Rate']=LE.fit_transform(df['Per_Km_Rate'])
df['Per_Minute_Rate']=LE.fit_transform(df['Per_Minute_Rate'])
df['Trip_Duration_Minutes']=LE.fit_transform(df['Trip_Duration_Minutes'])
```

```python
from sklearn.preprocessing import OrdinalEncoder
Or = OrdinalEncoder()
df[['Traffic_Conditions']] = Or.fit_transform(df[['Traffic_Conditions']])
```

```python
x=df.drop(columns='Trip_Price')
x
```

```python
y=df['Trip_Price']
y
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.30, random_state=42)
```
